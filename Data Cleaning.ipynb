```python
#import library

import pandas as pd 
```

```python
#read excel file

df = pd.read_excel(r"C:\Users\Varsha\Downloads\Customer Call List.xlsx")
df
```

```python
#drop duplicates

df = df.drop_duplicates()
df
```

```python
#Drop the unnecessary columns

df = df.drop(columns ="Not_Useful_Column", axis = 1)
df
```

```python
#removing the ..., /, _ from the dataframe

df["Last_Name"] = df["Last_Name"].str.strip("./_")
df
```

```python
df["Phone_Number"] = df["Phone_Number"].str.replace('[^a-zA-Z0-9]','', regex=True)
df
```

```python
#format the phone number 
df["Phone_Number"] = df["Phone_Number"].apply(lambda x:str(x))
df["Phone_Number"] = df["Phone_Number"].apply(lambda x:x[0:3]+'-'+x[3:6]+'-'+x[6:10])
df
```

```python
#replace the nan-- and Na-- values 

df = df.replace(["nan--", "Na--"], "")
df
```

```python
df["Address"].str.split(',',n=2,expand=True)
```

```python
df[["street_address","state","zip_code"]] = df["Address"].str.split(" ",n=2,expand=True)
df
```

```python
df = df.drop(columns ="Address", axis = 1)
df
```

```python
df["Paying Customer"] = df["Paying Customer"].replace("Yes","Y")
```

```python
df["Paying Customer"] = df["Paying Customer"].replace("Y","Yes")
df
```

```python
df["Paying Customer"] = df["Paying Customer"].replace("No","N")
df["Paying Customer"] = df["Paying Customer"].replace("N","No")
df
```

```python
df["Paying Customer"] = df["Paying Customer"].replace("N/a","")
df
```

```python
df["Do_Not_Contact"] = df["Do_Not_Contact"].replace({"No":"N","Yes":"Y"})
df["Do_Not_Contact"] = df["Do_Not_Contact"].replace({"N":"No","Y":"Yes"})
df
```

```python
df = df.fillna("")
```

```python
df
```

```python
for x in df.index:
    if df.loc[x, "Do_Not_Contact"] == 'Yes':
        df.drop(x, inplace= True)
df
    
```

```python
for x in df.index:
    if df.loc[x, "Phone_Number"] == '':
        df.drop(x, inplace= True)
df
    
```

```python
df=df.reset_index(drop=True)
df
```
